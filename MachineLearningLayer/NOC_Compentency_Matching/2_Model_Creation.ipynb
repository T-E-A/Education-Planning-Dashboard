```python
import pandas as pd 
import json
```

```python
with open('Compentencies_Features.json', 'r') as f:
    Feature_List = json.load(f)
```

```python
Feature_List
```

```python
df_Data = pd.read_csv('NOC_Competencies.csv')
```

```python
df_Data
```

```python
Target = ['OaSIS 2021 Labels']
All_Columns = Feature_List + Target
df_Data_Reduced = df_Data[All_Columns]
```

```python
df_Data_Reduced = df_Data_Reduced.rename(columns={'OaSIS 2021 Labels':'Target'})
print(df_Data_Reduced)
```

Generating Simple Nearest Neighbour Model

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
```

```python
competencies = df_Data_Reduced.drop(columns=["Target"])
jobs = df_Data_Reduced["Target"]
```

```python
scaler = StandardScaler()
competencies_scaled = scaler.fit_transform(competencies)
```

```python
nn = NearestNeighbors(
    n_neighbors=7,
    metric='cosine'   
)

nn.fit(competencies_scaled)
```

```python
def recommend_jobs(user_input):

    user_scaled = scaler.transform([user_input])

    distances, indices = nn.kneighbors(user_scaled)

    return jobs.iloc[indices[0]].tolist()
```

```python
user = [5,2,2,5,3,5,4,4,4,4,4]

recommend_jobs(user)
```

Saving the Model 

```python
import joblib
```

```python
joblib.dump({
    "model": nn,
    "scaler": scaler,
    "jobs": jobs,
    "X_scaled": competencies_scaled
}, "career_matcher.pkl")
```

```python
jobs.to_csv('jobs.csv')
```

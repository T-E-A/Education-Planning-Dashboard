```python
import pandas as pd 
from scipy.stats import ttest_rel
from scipy.stats import wilcoxon
import matplotlib.pyplot as plt
```

Comparing Models based on Mean Absolute Percentage Error 


Comparing Prophet vs. XGB

```python
prophet_path = '../Prophet/Data/Evaluation_Metrics.csv'
xgb_path = '../XGB/Data/Evaluation_Metrics.csv'
```

```python
df_Prophet = pd.read_csv(prophet_path)
df_XGB = pd.read_csv(xgb_path)
```

```python
df_Prophet_Reduced = df_Prophet[['provinceid','noc_groupingid','CV RMSE']].rename(columns={'CV RMSE':'Prophet CV RMSE'})
```

```python
df_XGB_Reduced = df_XGB[['provinceid','noc_groupingid','CV RMSE']].rename(columns={'CV RMSE':'XGB CV RMSE'})
```

```python
df_prophet_vs_xgb = df_Prophet_Reduced.merge(df_XGB_Reduced, on=['noc_groupingid','provinceid'])
df_prophet_vs_xgb
```

```python
ax = (
    df_prophet_vs_xgb['Prophet CV RMSE']
    - df_prophet_vs_xgb['XGB CV RMSE']
).hist(bins=30, figsize=(10, 6))

ax.set_title('Distribution of CV RMSE Differences\n(Prophet − XGB)')
ax.set_xlabel('CV RMSE Difference')
ax.set_ylabel('NOC and Province Groupings')

ax.axvline(0, linestyle='--')

plt.show()
```

```python
#Parametric Test 

# Null hypothesis (H0):
#   The average CV RMSE of Prophet is GREATER THAN OR EQUAL TO XGB.
#   mean(Prophet RMSE − XGB RMSE) > 0
#
# Alternative hypothesis (H1):
#   Prophet has a LOWER average CV RMSE than XGB.
#   mean(Prophet RMSE − XGB RMSE) < 0



t_stat, p_val = ttest_rel(df_prophet_vs_xgb['Prophet CV RMSE'], df_prophet_vs_xgb['XGB CV RMSE'], alternative='less')

# The t-statistic indicates direction and magnitude:
#   - Negative value --> Prophet tends to have lower RMSE
#   - Larger magnitude --> stronger average difference
print(f"T-Stat:\n\t{t_stat}")

# The p-value tells us how likely it is to observe this result
# if Prophet were NOT better than XGB.
# A significantly small p-value (e.g. < 0.05) tells us that Prophet has significantly less error.
print(f"P-Value:\n\t{p_val}")
```

```python
#Non Parametric Test

# Null hypothesis (H0):
#   The median CV RMSE difference is GREATER THAN OR EQUAL TO zero.
#   median(Prophet RMSE − XGB RMSE) >= 0
#
# Alternative hypothesis (H1):
#   Prophet has LOWER CV RMSE in the typical forecasting task.
#   median(Prophet RMSE − XGB RMSE) < 0


wilcoxon_result = wilcoxon(
    df_prophet_vs_xgb['Prophet CV RMSE'],
    df_prophet_vs_xgb['XGB CV RMSE'],
    alternative='less'
)

# The p-value:
# Probability of observing the consistency of which Prophets CV RMSE is lower than XGBs CV RMSE. 
print(f"P-Value:\n\t{wilcoxon_result.pvalue}")
```

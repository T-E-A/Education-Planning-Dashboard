Pre Processing - Defining NOC Grouping Identifers for TS forecasting

```python
'''
The purpose of this notebook is to determine the identifiers for the NOC Grouping and the 
occupations that were grouped into them.
'''
```

```python
import pandas as pd 
```

```python
df_Dimension = pd.read_csv('metadata/14100421_MetaData_Dimensions.csv')
df_Dimension
```

```python
#We are interested in Dimension 3 

df_Members = pd.read_csv('metadata/14100421_MetaData_Members.csv')
df_Members
```

```python
df_Members_Reduced = df_Members[df_Members['Dimension ID']==3][['Member Name','Member ID','Member Notes']]
df_Members_Reduced
```

```python
df_Notes = pd.read_csv('metadata/14100421_MetaData_Notes.csv')
df_Notes
```

```python
NoteID_Dict = df_Notes.set_index('Note ID')['Note'].to_dict()
```

```python
df_Members_Reduced['Grouping Notes'] = df_Members_Reduced['Member Notes'].map(NoteID_Dict)
df_Members_Reduced
```

```python
df_Final_Tbl_NOC_Grouping_ID = df_Members_Reduced[['Member Name', 'Member ID', 'Grouping Notes']].rename(columns=
{'Member Name':'NOC_Grouping',
 'Member ID': 'NOC_GroupingID',  
})

df_Final_Tbl_NOC_Grouping_ID
```

```python
df_Final_Tbl_NOC_Grouping_ID['Grouping Notes'] = df_Final_Tbl_NOC_Grouping_ID['Grouping Notes'].fillna('No Notes')
df_Final_Tbl_NOC_Grouping_ID
```

```python
df_Final_Tbl_NOC_Grouping_ID.to_csv('processed_data/Tbl_NOC_TS_Groupings.csv', index=False)
```

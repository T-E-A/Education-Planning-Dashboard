```python

df_Data = pd.read_csv('Tbl_NOC_Competencies.csv')

```

```python
Target = ['OaSIS 2021 Labels']
All_Columns = Feature_List + Target
df_Data_Reduced = df_Data[All_Columns]
```

```python
df_Data_Reduced = df_Data_Reduced.rename(columns={'OaSIS 2021 Labels':'Target'})
print(df_Data_Reduced)
```

Generating Simple Nearest Neighbour Model

```python
from sklearn.neighbors import NearestNeighbors
from sklearn.preprocessing import StandardScaler
```

```python
competencies = df_Data_Reduced.drop(columns=["Target"])
jobs = df_Data_Reduced["Target"]
```

```python
scaler = StandardScaler()
competencies_scaled = scaler.fit_transform(competencies)
```

```python
nn = NearestNeighbors(
    n_neighbors=7,
    metric='euclidean'   
)

nn.fit(competencies_scaled)
```

```python
def recommend_jobs(user_input):

    user_scaled = scaler.transform([user_input])

    distances, indices = nn.kneighbors(user_scaled)

    results = jobs.iloc[indices[0]].reset_index(drop=True)
    distances = distances[0]

    output = results.to_frame(name="NOC Occupation")
    output["Distance"] = distances

    return output
```

```python
user = [1,3,2,0,2,1,1,0,2,2,1]

recommended_jobs = recommend_jobs(user)
```

```python
ds_Sample = recommended_jobs.merge(df_Data_Reduced.rename(columns={'Target':'NOC Occupation'}), on=['NOC Occupation'])
```

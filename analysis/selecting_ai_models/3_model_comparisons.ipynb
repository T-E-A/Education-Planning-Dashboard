```python
df_Results["Consistency Label"] = (
    df_Results[["NEUTRAL", "ENTAILMENT", "CONTRADICTION"]]
    .idxmax(axis=1)
    .str.title()
)

df_Results
```

```python
df_Entailment_Instances = df_Results[df_Results['Consistency Label']=='Entailment']
df_Entailment_Instances
```

```python
df_Entailment_Distributions = df_Entailment_Instances.groupby(['Model'])['ENTAILMENT'].describe().round(3).reset_index()
df_Entailment_Distributions
```

```python
Entailment_Scatter = px.bar(
    title='Number of Entailments vs. Avg Confidence in Label',
    data_frame=df_Entailment_Distributions,
    x='Model',
    y='count',
    color='Model',
    color_discrete_map=Model_Discrete_Map,
    width=1500,
    height=1000,
    template='ggplot2',
    text='count'
)

Entailment_Scatter.add_trace(
    go.Scatter(
        x=df_Entailment_Distributions['Model'],
        y=df_Entailment_Distributions['mean'],
        mode='lines+markers',
        marker=dict(color='#474545', size=18),
        line=dict(color="#1A1919", width=2.5),
        name='Avg Confidence in Label',
        yaxis='y2'   # <-- Secondary axis
    )
)

Entailment_Scatter.update_layout(
    title_font=dict(size=16),
    font=dict(size=12),

    hoverlabel=dict(font_size=20),

    legend=dict(
        orientation='h',
        yanchor='bottom',
        y=-0.3,
        xanchor='center',
        x=0.5,
        font=dict(size=16)
    ),

    plot_bgcolor="#CDD0D4",

    xaxis=dict(
        title="Model",
        title_font=dict(size=14),
        tickfont=dict(size=12)
    ),

    yaxis=dict(
        title="Entailment Count",
        title_font=dict(size=14),
        tickfont=dict(size=12)
    ),

    yaxis2=dict(
        title="Avg Confidence",
        title_font=dict(size=14),
        tickfont=dict(size=12),
        overlaying='y',   # share same x-axis
        side='right'
    )
)

Entailment_Scatter.show()
```

```python
Entailment_Scatter.write_image("Figures/Entailment_Count_vs_Mean_Confidence.svg")
```

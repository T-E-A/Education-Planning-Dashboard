```python
import json
```

```python
noc_path = 'DataSets/ground_truth_data_noc.json'
fos_path = 'DataSets/ground_truth_data_FOSinfo.json'
mhs_path = 'DataSets/ground_truth_data_MHS.json'
lf_path = 'DataSets/ground_truth_data_labourforce.json'
gs_path = 'DataSets/ground_truth_data_graduatesurvey.json'
```

```python
paths = [noc_path, fos_path, mhs_path, lf_path, gs_path]
```

```python
Full_Data_Set = []

for path in paths:
    with open(path, 'r') as f:
        data = json.load(f)
    
    Full_Data_Set.extend(data)
```

```python
len(Full_Data_Set)
```

```python
idx = 1

for i in range(len(Full_Data_Set)):
    Full_Data_Set[i]['Prompt_ID'] = idx
    idx += 1
```

```python
Full_Data_Set
```

```python
with open("DataSets/Full_Ground_Truth_Data_Set.json", "w", encoding='utf-8') as f:
    json.dump(Full_Data_Set, f, indent=4, ensure_ascii=False)
```
